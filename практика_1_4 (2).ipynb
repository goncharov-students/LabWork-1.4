Задание 1.1. Базовые свойства массивов

Вы анализируете данные о заказах интернет-магазина.

Для каждого заказа известны:

• время доставки в минутах;

• стоимость заказа в рублях;

• количество товаров в заказе.

Создайте массивы:

import numpy as np

delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])

order_price = np.array([1200, 3500, 800, 4200, 1900, 5600,
1500, 2800])

items_count = np.array([2, 5, 1, 7, 3, 8, 2, 4])

Что необходимо сделать
1. Для каждого массива выведите:
◦ shape;
◦ ndim;
◦ size.
2. Объясните, что означает size в контексте рассматриваемых данных.

size — это количество наблюдений (заказов) в массиве. В данном случае в каждом массиве по 8 заказов.

3. Проверьте с помощью Python, что все три массива содержат одинаковое количество
наблюдений.

4. Объясните, почему каждый индекс в трёх массивах должен соответствовать одному и
тому же заказу.

Потому что массивы описывают разные характеристики одних и тех же объектов. Если порядок будет нарушен, мы получим несогласованные данные (например, время доставки от одного заказа, а стоимость — от другого). Это приведёт к неверным выводам.

Вопрос для анализа

Что произойдёт с результатами анализа, если delivery_time содержит 8 наблюдений, а
order_price — только 7? Почему это является проблемой не только с точки зрения
Python, но и с точки зрения качества данных?

Если delivery_time содержит 8 наблюдений, а order_price — 7, то:

Python не сможет корректно выполнять поэлементные операции (например, delivery_time + order_price вызовет ошибку).

С точки зрения качества данных это означает потерю или несогласованность информации: неизвестно, какому заказу соответствует пропущенная стоимость. Анализ становится ненадёжным.



```python
import numpy as np
delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])
order_price = np.array([1200, 3500, 800, 4200, 1900, 5600, 1500, 2800])
items_count = np.array([2, 5, 1, 7, 3, 8, 2, 4])
# 1. Вывод shape, ndim, size для каждого массива
for name, arr in [("delivery_time", delivery_time),
                  ("order_price", order_price),
                  ("items_count", items_count)]:
    print(f"{name}: shape={arr.shape}, ndim={arr.ndim}, size={arr.size}")
# 3. Проверка одинакового количества наблюдений
print("\nОдинаковое количество наблюдений:",
      delivery_time.size == order_price.size == items_count.size)
```

orders = np.array([

Объясните, почему:

orders.shape
возвращает (8, 3), а:
orders.size
возвращает 24.

Что характеризует каждый из этих показателей?

orders.shape возвращает (8, 3), потому что в матрице 8 строк (заказов) и 3 столбца (признака).
orders.size возвращает 24, потому что это общее количество элементов: 8 × 3 = 24.

Выделите из матрицы orders столбец времени доставки:

delivery = orders[:, 0]

Что необходимо сделать
1. Измените первый элемент delivery на 1000.

4. Восстановите исходную матрицу orders.
5. Создайте независимую копию:
delivery_copy = orders[:, 0].copy()
6. Измените первый элемент delivery_copy.
7. Снова выведите orders.
8. Сравните результаты двух экспериментов.

Вопрос для анализа

Почему понимание различия между view и copy важно при обработке реальных наборов
данных?

Приведите пример ситуации, в которой случайное изменение view может привести к ошибке
в дальнейшем анализе.
Вывод:
В первом эксперименте изменение delivery повлияло на orders, потому что срез orders[:, 0] возвращает view (представление), а не новый массив. View ссылается на те же данные в памяти.

Во втором эксперименте delivery_copy — это независимая копия, поэтому изменение не затрагивает исходную матрицу.

Ответ на вопрос для анализа:

Понимание разницы между view и copy важно, потому что случайное изменение view может незаметно исказить исходные данные. Например, при очистке данных можно случайно изменить столбец в исходной матрице, что приведёт к ошибкам в дальнейшем анализе.

Пример: вы нормализуете столбец delivery, думая, что создаёте новый массив, но на самом деле меняете исходные данные, и все последующие расчёты становятся неверными.



```python
# Восстановим исходную матрицу
orders = np.array([
    [35, 1200, 2],
    [42, 3500, 5],
    [28, 800, 1],
    [51, 4200, 7],
    [39, 1900, 3],
    [63, 5600, 8],
    [31, 1500, 2],
    [45, 2800, 4]
])

# 1. Выделяем столбец (view)
delivery = orders[:, 0]
delivery[0] = 1000
print("orders после изменения view:\n", orders)

# 4. Восстанавливаем исходную матрицу
orders = np.array([
    [35, 1200, 2],
    [42, 3500, 5],
    [28, 800, 1],
    [51, 4200, 7],
    [39, 1900, 3],
    [63, 5600, 8],
    [31, 1500, 2],
    [45, 2800, 4]
])

# 5. Создаём независимую копию
delivery_copy = orders[:, 0].copy()
delivery_copy[0] = 1000
print("orders после изменения copy:\n", orders)
```

Маска [False, False, False, True, False, True, False, False] показывает, что проблемными являются 4-й и 6-й заказы (51 и 63 минуты). Среднее время для них — 57 минут.

Задание 1.5. Фильтрация по нескольким условиям

Менеджер продукта хочет изучить заказы, которые одновременно являются крупными по
стоимости и содержат большое количество товаров.

Критерии:

• стоимость заказа больше 3000 рублей;

• количество товаров больше 4.

Что необходимо сделать
1. Создайте единую булеву маску, учитывающую оба условия.
2. Получите стоимость подходящих заказов.
3. Получите время доставки этих заказов.
4. Рассчитайте среднее время доставки выбранной группы.

Вопрос для анализа

Почему анализ комбинации нескольких признаков может дать более полезный результат для
бизнеса, чем анализ каждого признака отдельно?

Приведите пример ситуации, когда отдельный показатель выглядит нормально, но комбинация
нескольких признаков выявляет проблему.
Под условия подходят заказы с индексами 1, 3, 5 (3500/5, 4200/7, 5600/8). Среднее время доставки — (42 + 51 + 63) / 3 = 52 минуты.

Ответ на вопрос для анализа:

Анализ комбинации признаков даёт более полезный результат, потому что позволяет выявить сегменты, которые не видны при одномерном анализе. Например, заказ может быть крупным по стоимости, но с малым количеством товаров — это не проблема. А комбинация «дорогой + много товаров» может указывать на сложные заказы, которые чаще задерживаются.

Пример: отдельно время доставки может быть в норме, но в сочетании с большим количеством товаров и высокой стоимостью оно резко возрастает — это сигнал для оптимизации логистики.



```python
order_price = np.array([1200, 3500, 800, 4200, 1900, 5600, 1500, 2800])
items_count = np.array([2, 5, 1, 7, 3, 8, 2, 4])
delivery_time = np.array([35, 42, 28, 51, 39, 63, 31, 45])
# 1. Единая булева маска
mask = (order_price > 3000) & (items_count > 4)
print("Маска:", mask)
# 2. Стоимость подходящих заказов
print("Стоимость подходящих заказов:", order_price[mask])
# 3. Время доставки этих заказов
print("Время доставки подходящих заказов:", delivery_time[mask])
# 4. Среднее время доставки
print("Среднее время доставки:", delivery_time[mask].mean())
```

Задание 2.2. Процентили и SLA

Интернет-магазин заявляет:

Не менее 90% заказов доставляются за 40 минут или быстрее.

Для проверки SLA смоделируйте 1000 наблюдений времени доставки.

Используйте нормальное распределение со следующими параметрами:

• среднее — 30 минут;

• стандартное отклонение — 5 минут.

np.random.seed(42)

delivery_simulated = np.random.normal(

 loc=30,

 scale=5,

 size=1000

)

Что необходимо сделать

1. Рассчитайте:

◦ среднее значение;

◦ стандартное отклонение;

◦ медиану.

2. Рассчитайте следующие процентили:

p50

p90

p95

p99

используя np.quantile().

3. Сравните полученные значения с параметрами, использованными при генерации
данных.

4. Используя p90, определите, выполняется ли заявленный SLA.

Вопросы для анализа

1. Что означает p90 для времени доставки?

2. Что означает p95?

3. Почему для распределения времени доставки p95 обычно выше p90?

4. Что изменится в управлении логистикой, если компания начнёт ориентироваться не
только на p90, но и на p95?

Продуктовый вывод

Сформулируйте вывод для менеджера продукта в 2–3 предложениях:

• выполняется ли SLA;

• насколько хорошо контролируется длинный хвост распределения.
Вывод:
Среднее ≈ 30, std ≈ 5, медиана ≈ 30. p90 ≈ 36.4, p95 ≈ 38.2, p99 ≈ 41.6.
SLA выполняется, так как p90 ≤ 40.

Ответы на вопросы:

p90 — это значение, ниже которого находится 90% наблюдений. 90% заказов доставляются за время ≤ p90.

p95 — 95% заказов доставляются за время ≤ p95.

p95 выше p90, потому что он охватывает более длинный хвост распределения — самые медленные доставки.

Если компания начнёт ориентироваться на p95, ей придётся улучшать логистику для самых медленных 5% заказов, что может потребовать дополнительных ресурсов.

Продуктовый вывод:
SLA выполняется: 90% заказов доставляются за 40 минут или быстрее. Однако длинный хвост распределения (p95, p99) показывает, что есть заказы с существенно большим временем доставки, что требует внимания.

```python
np.random.seed(42)
delivery_simulated = np.random.normal(loc=30, scale=5, size=1000)
print("Среднее:", delivery_simulated.mean())
print("Стандартное отклонение:", delivery_simulated.std())
print("Медиана:", np.median(delivery_simulated))
p50 = np.quantile(delivery_simulated, 0.50)
p90 = np.quantile(delivery_simulated, 0.90)
p95 = np.quantile(delivery_simulated, 0.95)
p99 = np.quantile(delivery_simulated, 0.99)
print(f"p50={p50:.2f}, p90={p90:.2f}, p95={p95:.2f}, p99={p99:.2f}")
sla_met = p90 <= 40
print("SLA выполняется:", sla_met)
```

• среднее;

Вопросы для анализа

Объясните ответ.
Вывод:
r ≈ 0.95 — сильная положительная линейная связь. Направление: с увеличением количества товаров время доставки растёт. Сила связи: очень сильная.

Объяснение: Чем больше товаров в заказе, тем дольше его собирают на складе и тем больше времени занимает доставка.

Ответ на вопрос:
Нельзя утверждать, что увеличение количества товаров является причиной увеличения времени доставки, потому что корреляция не доказывает причинно-следственную связь. Возможны скрытые факторы (например, тип товара, удалённость клиента).

```python
np.random.seed(42)
items = np.random.randint(1, 10, 100)
delivery_time_correlated = 20 + 3 * items + np.random.normal(0, 3, 100)
corr_matrix = np.corrcoef(items, delivery_time_correlated)
r = corr_matrix[0, 1]
print(f"Коэффициент корреляции r = {r:.3f}")
```

Задание 3.1. Bootstrap и доверительный интервал

На складе измерили время сборки шести заказов:
x = np.array([28, 31, 29, 45, 32, 30])

Что необходимо сделать
1. Рассчитайте среднее значение исходной выборки.
2. Реализуйте Bootstrap с количеством итераций:
B = 5000
На каждой итерации:
• сформируйте выборку такого же размера, как x;
• используйте выборку с возвращением;
• рассчитайте её среднее;
• сохраните результат.
Для формирования Bootstrap-выборки используйте:
np.random.choice()
3. Получите центральный 95%-й доверительный интервал для среднего с помощью:
np.quantile()
Используйте квантили:
0.025
0.975
4. Выведите полученный доверительный интервал.

Вопросы для анализа

1. Почему Bootstrap используется именно с выборкой с возвращением?
2. Как наличие наблюдения 45 влияет на Bootstrap-распределение среднего?
3. Почему результаты Bootstrap могут заметно изменяться при использовании очень
маленькой исходной выборки?
4. Можно ли утверждать:
«Вероятность того, что истинное среднее находится внутри рассчитанного 95%-го
доверительного интервала, равна 95%»?
Если нет, сформулируйте корректную интерпретацию доверительного интервала.
Вывод:
Среднее исходной выборки ≈ 32.5. Доверительный интервал ≈ [29.0, 36.5].

Ответы на вопросы:

Bootstrap используется с возвращением, чтобы имитировать случайность выборки из генеральной совокупности.

Наблюдение 45 увеличивает разброс bootstrap-распределения и сдвигает интервал вправо.

При маленькой выборке bootstrap-распределение очень изменчиво, интервал может быть неточным.

Утверждение неверно. Корректная интерпретация: если бы мы повторяли эксперимент много раз, 95% таких интервалов накрывали бы истинное среднее.

```python
x = np.array([28, 31, 29, 45, 32, 30])
print("Среднее исходной выборки:", x.mean())
B = 5000
boot_means = np.empty(B)
for i in range(B):
    sample = np.random.choice(x, size=x.size, replace=True)
    boot_means[i] = sample.mean()
ci_lower = np.quantile(boot_means, 0.025)
ci_upper = np.quantile(boot_means, 0.975)
print(f"95% доверительный интервал: [{ci_lower:.2f}, {ci_upper:.2f}]")
```

◦ что означает ошибка II рода.

4. Укажите, какая ошибка соответствует:

◦ ложной тревоге;

◦ пропущенному эффекту.

Ситуация для анализа
Полноценная эксплуатация нового алгоритма обходится компании в 5 млн рублей в год.
